```python
import pandas as pd
import numpy as np

# Carregar o dataset
df = pd.read_csv('conjunto_dados_psicologia_25k.csv')

# Criar colunas auxiliares/derivadas para as análises
df['Tempo_Tela_Horas'] = (df['Tempo_Tela_Min'] / 60).round(2)
df['Faixa_Etaria'] = pd.cut(df['Idade'], bins=[17, 30, 45, 60, 100], labels=['18-30', '31-45', '46-60', '60+'])

print("Dados carregados com sucesso!")
```

```python
# Pergunta 1
print("PERGUNTA 1: Qual é a distribuição percentual dos Diagnósticos Preliminares na base de dados?\n")

p1 = df['Diagnostico_Preliminar'].value_counts(normalize=True) * 100

print("RESPOSTA 1:")
print(p1.round(2).to_string())
```

```python
# Pergunta 2
print("PERGUNTA 2: Qual é a média geral de Horas de Sono e de Tempo de Tela (convertido para horas)?\n")

media_sono = df['Horas_Sono'].mean()
media_tela = df['Tempo_Tela_Horas'].mean()

print("RESPOSTA 2:")
print(f"Média de Horas de Sono: {media_sono:.2f} horas")
print(f"Média de Tempo de Tela: {media_tela:.2f} horas")
```

```python
# Pergunta 3
print("PERGUNTA 3: Como varia a média de Horas de Sono entre cada categoria de Diagnóstico Preliminar?\n")

p3 = df.groupby('Diagnostico_Preliminar')['Horas_Sono'].mean().sort_values()

print("RESPOSTA 3:")
print(p3.round(2).to_string())
```

```python
# Pergunta 4
print("PERGUNTA 4: Qual Mecanismo de Enfrentamento apresenta a maior média no Nível de Estresse?\n")

p4 = df.groupby('Mecanismo_Enfrentamento')['Nivel_Estresse'].mean().sort_values(ascending=False)

print("RESPOSTA 4:")
print(p4.round(2).to_string())
```

```python
# Pergunta 5
print("PERGUNTA 5: Qual é a pontuação média do PHQ-9 (Depressão) e do GAD-7 (Ansiedade) dividida por Gênero?\n")

p5 = df.groupby('Genero')[['Pontuacao_PHQ9', 'Pontuacao_GAD7']].mean()

print("RESPOSTA 5:")
print(p5.round(2))
```

```python
# Pergunta 6
print("PERGUNTA 6: Quantos pacientes possuem indicadores de rastreio positivo moderado/grave (PHQ-9 >= 10 e GAD-7 >= 10 simultaneamente)?\n")

p6 = len(df[(df['Pontuacao_PHQ9'] >= 10) & (df['Pontuacao_GAD7'] >= 10)])
porcentagem_p6 = (p6 / len(df)) * 100

print("RESPOSTA 6:")
print(f"Total de pacientes com PHQ-9 >= 10 e GAD-7 >= 10: {p6} ({porcentagem_p6:.2f}% do total)")
```

```python
# Pergunta 7
print("PERGUNTA 7: Qual é a média de idade dos pacientes para cada tipo de Diagnóstico Preliminar?\n")

p7 = df.groupby('Diagnostico_Preliminar')['Idade'].mean().sort_values()

print("RESPOSTA 7:")
print(p7.round(1).to_string())
```

```python
# Pergunta 8
print("PERGUNTA 8: Qual o percentual de pacientes que utilizam 'Isolamento' ou 'Redes Sociais' como mecanismo de enfrentamento?\n")

mecanismos = df['Mecanismo_Enfrentamento'].isin(['Isolamento', 'Redes Sociais'])
p8 = (mecanismos.sum() / len(df)) * 100

print("RESPOSTA 8:")
print(f"Percentual de pacientes que utilizam Isolamento ou Redes Sociais: {p8:.2f}%\n")

print("ANÁLISE INTERPRETATIVA 8:")
print(f"Observa-se que {p8:.2f}% dos pacientes utilizam estratégias de enfrentamento focadas em esquiva "
      f"ou hábitos digitais (Isolamento/Redes Sociais). Esse volume expressivo sugere uma tendência "
      f"de comportamento passivo diante de quadros de estresse e ansiedade, o que pode agravar o quadro clínico "
      f"a longo prazo por falta de suporte social direto.")
```

```python
# Pergunta 9
print("PERGUNTA 9: Como se comporta o tempo médio de tela quando agrupado por faixa etária (18-30, 31-45, 46-60, 60+)?\n")

p9 = df.groupby('Faixa_Etaria', observed=False)['Tempo_Tela_Horas'].mean()

print("RESPOSTA 9:")
print(p9.round(2).to_string())
print("\nANÁLISE INTERPRETATIVA 9:")
faixa_maior = p9.idxmax()
maior_tempo = p9.max()
print(f"A análise por faixa etária indica que a faixa de '{faixa_maior}' lidera o consumo digital "
      f"com média de {maior_tempo:.2f} horas diárias. O padrão demonstra que públicos mais jovens "
      f"possuem maior exposição continuada a telas, o que se relaciona diretamente com níveis mais elevados "
      f"de fadiga mental e queixas de insônia relatadas na amostra.")
```

```python
# Pergunta 10
print("PERGUNTA 10: Quem são os 5 pacientes com maior tempo de tela que possuem diagnósticos graves?\n")

diagnosticos_graves = ['Sintomas Depressivos Graves', 'Ansiedade Grave', 'Depressão e Ansiedade Graves']
p10 = df[df['Diagnostico_Preliminar'].isin(diagnosticos_graves)].nlargest(5, 'Tempo_Tela_Min')[
    ['ID_Paciente', 'Idade', 'Tempo_Tela_Horas', 'Diagnostico_Preliminar']
]

print("RESPOSTA 10:")
print(p10.to_string(index=False))
print("\nANÁLISE INTERPRETATIVA 10:")
print("Ao filtrar os casos críticos (diagnósticos graves), identifica-se um padrão marcante de "
      "tempo de tela extremamente elevado nestes 5 indivíduos. Isso reforça a hipótese da literatura de que "
      "o tempo excessivo em telas atua como um fator correlacionado ou de agravamento em quadros graves "
      "de saúde mental, exigindo intervenções focadas na higiene do sono e desintoxicação digital.")
```
